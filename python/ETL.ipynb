```python
import pandas as pd
import sqlite3
import seaborn as sns
import matplotlib.pyplot as plt

path = '../portfolio.db'
conn = sqlite3.connect(path)

cur = conn.cursor()
```

### Extract from the file database

```python
sales_df = pd.read_sql("select * from Superstore", conn)
sales_df.head(10)
```

### Transform

```python
print(sales_df.info())
```

```python
sales_df['OrderDate'] = pd.to_datetime(sales_df['OrderDate'], format='%m/%d/%Y')
print(sales_df.head(10))
```

```python
sales_df['ShipDate'] = pd.to_datetime(sales_df['ShipDate'], format='%m/%d/%Y')
print(sales_df.head(10))
```

```python
sales_df.head()
```

```python

sales_df['Sales'] = sales_df['Sales'].round(2)
sales_df['Profit'] = sales_df['Profit'].round(2)
print(sales_df.head())
```

```python
sales_df.describe(exclude=['str'])
```

```python
# check for date inconsistency
print("checking for date inconsistencies...")
inconsistency = sales_df[sales_df['OrderDate']>sales_df['ShipDate']][['OrderID','OrderDate','ShipDate']]

if inconsistency.isnull:
    print("All order dates are consistent (OrderDate < ShipDate)")
else:
    print("Date consistency is detected:")
    print(inconsistency)
```

```python
import kagglehub
from kagglehub import KaggleDatasetAdapter

# Set the path to the file you'd like to load
file_path = "zip_lat_long.csv"

# Load the latest version
df = kagglehub.load_dataset(
  KaggleDatasetAdapter.PANDAS,
  "joeleichter/us-zip-codes-with-lat-and-long",
  file_path,
  # Provide any additional arguments like 
  # sql_query or pandas_kwargs. See the 
  # documenation for more information:
  # https://github.com/Kaggle/kagglehub/blob/main/README.md#kaggledatasetadapterpandas
)

print("First 5 records:", df.head())
```

```python
# Make the column names consistent
df.columns = ['ZipCode', 'Latitude', 'Longitude']
```

```python
# Merge sales data with zip code data
merged_df = pd.merge(sales_df, df, left_on='PostalCode', right_on='ZipCode', how='left')
print("Merged DataFrame head:", merged_df.head())
```
